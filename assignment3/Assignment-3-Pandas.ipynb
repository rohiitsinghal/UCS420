### Q.6 Employees dataset

```python
#Assignment 6: WAP to create a sample dataset (employees.csv) containing information about employees in a company.
emp_data = {
    "Employee_ID": [101,102,103,104,105],
    "Name": ["Alice","Bob","Charlie","Diana","Edward"],
    "Department": ["HR","IT","IT","Marketing","Sales"],
    "Age": [29,34,41,28,38],
    "Salary": [50000,70000,65000,55000,60000],
    "Years_of_Experience": [4,8,10,3,12],
    "Joining_Date": ["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-25"],
    "Gender": ["Female","Male","Male","Female","Male"],
    "Bonus": [5000,7000,6000,4500,5000],
    "Rating": [4.5,4.0,3.8,4.7,3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv("employees.csv", index=False)
emp_df
```

```python
#a) Shape (number of rows and columns) of the DataFrame.
emp_df.shape
```

```python
#b) Summary of the DataFrame that includes the data types and non-null counts for each column.
emp_df.info()
```

```python
#c) Generate descriptive statistics.
emp_df.describe()
```

```python
#d) Display the first 5 rows and last 3 rows of the dataset.
print("First 5 rows:")
display(emp_df.head())
print("Last 3 rows:")
display(emp_df.tail(3))
```

```python
#e) i. The average salary of employees.
avg_salary = emp_df["Salary"].mean()
print("Average Salary:", avg_salary)
```

```python
#e) ii. The total bonus paid to all employees.
total_bonus = emp_df["Bonus"].sum()
print("Total Bonus:", total_bonus)
```

```python
#e) iii. The youngest employee's age.
youngest_age = emp_df["Age"].min()
print("Youngest Employee's Age:", youngest_age)
```

```python
#e) iv. The highest performance rating.
highest_rating = emp_df["Rating"].max()
print("Highest Performance Rating:", highest_rating)
```

```python
#f) Sort the DataFrame by the Salary column in descending order.
emp_df_sorted = emp_df.sort_values(by="Salary", ascending=False)
emp_df_sorted
```

```python
#g) Add a new column that categorizes employees based on their performance rating.
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp_df["Performance_Category"] = emp_df["Rating"].apply(categorize)
emp_df
```

```python
#h) Identify missing values in the DataFrame.
emp_df.isnull().sum()
```

```python
#i) Rename the Employee_ID column to ID.
emp_df.rename(columns={"Employee_ID":"ID"}, inplace=True)
emp_df
```

```python
#j) i. Employees with more than 5 years of experience.
emp_df[emp_df["Years_of_Experience"] > 5]
```

```python
#j) ii. Employees who belong to the IT department.
emp_df[emp_df["Department"] == "IT"]
```

```python
#k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.
emp_df["Tax"] = emp_df["Salary"] * 0.10
emp_df
```

```python
#l) Save the modified DataFrame (with added columns) to a new CSV file.
emp_df.to_csv("employees_modified.csv", index=False)
print("File saved as employees_modified.csv")
```
