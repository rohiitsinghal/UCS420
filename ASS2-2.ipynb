## 3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers

```python
import random

random.seed(1024170221)
```

### i. Generate a list of 100 random numbers between 100 and 900 (inclusive).

```python
random_numbers = [random.randint(100, 900) for i in range(100)]

print(random_numbers)
```

### ii. Count and print all odd numbers in the list.

```python
odd_numbers = [x for x in random_numbers if x % 2 != 0]

print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

### iii. Count and print all even numbers in the list.

```python
even_numbers = [x for x in random_numbers if x % 2 == 0]

print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))
```

### iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

```python
def is_prime(n):
    if n < 2:
        return False
    
    for i in range(2,n):
        if n % i == 0:
            return False
    
    return True


prime_numbers = [x for x in random_numbers if is_prime(x)]

print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

### v. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
frequent_no=random_numbers[0]
c=random_numbers.count(frequent_no)
for i in random_numbers:
    s=random_numbers.count(i)
    if s>c:
        frequent_no=i
        c=s
print("Most frequent Number= ",frequent_no)
print("count= ",c)
```

## 4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

```python
n="1024170221"
l=[int(i) for i in n]
print(l)
```

```python
A={i *7 for i in l[:8]}
print(A)
```

```python
B={i *9 for i in l[:8]}
print(B)
```

### i. Find the union of A and B — all unique values across both sets.

```python
union_AB = A.union(B)

print("Union of A and B:", union_AB)
```

### ii. Find the intersection of A and B — values common to both.

```python
intersection_AB = A.intersection(B)

print("Intersection of A and B:", intersection_AB)
```

### iii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this differs from symmetric_difference().

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() finds elements in one set but not the other separately,
# while symmetric_difference() combines the elements unique to both sets.
```

### iv. Find the symmetric difference of A and B.

```python
symmetric_difference_AB = A.symmetric_difference(B)

print("Symmetric difference of A and B:", symmetric_difference_AB)
```

### v. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

### vi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain in one line why discard() is safer than remove() when you're not sure the value exists.

```python
X = int(input("Enter a value X: "))

A.discard(X)

print("Set A after discard():", A)

# discard() is safer than remove() because it does not raise an error
# when the specified value is not present in the set.
```
